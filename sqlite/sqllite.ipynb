SQL AND SQLITE
 structured Query Language is a standard language for managing

and manipulating relational databases, and SQLite is a self-contained, serverless and zero configuration

database engine that is widely used for embedded data systems.


```python
import sqlite3
```

```python
#connect to an SQLite db
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
  #create the table
cursor.execute('''Create Table If Not Exists emp(
               id Integer Primary Key,name Text Not Null,age Integer,department text)''')
```

```python
#commit changes
connection.commit()
```

```python
  #create the table
cursor.execute('''select* from emp''')
```

```python
#insert the data 
cursor.execute('''Insert into emp (name,age,department) values('bob',33,'engineering')''')
cursor.execute('''Insert into emp (name,age,department) values('jay',22,'engineering')''')
cursor.execute('''Insert into emp (name,age,department) values('krrish',43,'doctor')''')
```

```python
connection.commit()
```

```python
cursor.execute('select * from emp')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
#update the data
cursor.execute('''update emp set age=99 where name='krish'
               ''')
```

```python
cursor.execute('select * from emp')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute('''delete from emp where name='bob' ''')
connection.commit()
```

```python
#working iwth sales data
connection=sqlite3.connect('sales.db')
cursor=connection.cursor()
cursor.execute('''
create table if not exists sales (
               id integer primary key,
               date text not null,
               product text not null,
               sales integer,
               region text
               )
''')
sales=[
    ('2023-01-01','p1',100, 'north'),
    ('2023-02-01','p2',400, 'north'),
    ('2023-03-01','p3',1500, 'south'),
    ('2023-04-01','p4',1060, 'south')

]
cursor.executemany('''
insert into sales(date,product,sales,region)  
                   values (?,?,?,?)
                   ''', sales)
connection.commit()
```

```python
cursor.execute('select * from sales')
rows=cursor.fetchall()

for row in rows:
    print(row)
```
