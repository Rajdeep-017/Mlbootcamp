```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,2,3,5,7]
plt.plot(x,y)
plt.xlabel("x axia")
plt.ylabel("y axis")
plt.grid(True)
```

```python
x=[1,2,3,4,5]
y=[1,2,3,5,7]
#customised line plot
plt.plot(x,y,color='red',linestyle='--',marker="o",linewidth=2,markersize=5)
```

```python
#multiple Plot
x=[1,2,3,4,5]
y1=[1,2,4,6,8]
y2=[3,4,5,7,9]
plt.figure(figsize=(9,5))
plt.subplot(2,2,1) #row ,col,place of plot
plt.plot(x,y1,color='blue')
plt.plot()
plt.subplot(2,2,2) #row ,col,place of plot
plt.plot(x,y2,color='red')
plt.plot()
plt.subplot(2,2,3) #row ,col,place of plot
plt.plot(y1,y2,color='green')
plt.plot()
```

```python
#bar plot
cat=['a','b','c','d','e']
val=[1,2,3,4,5]
#barplot
plt.bar(cat,val,color="yellow")
plt.xlabel("categories")
```

HISTOGRAM
are use to represent the distribution of dataset .they divides data into bins and count the no of data points in each bin

```python
#histogram
val=[1,2,3,4,5,5,6,4,3,2,1]
plt.hist(val,bins=5,edgecolor='black')
```

```python
#scatter plot
x=[1,2,3,4,5]
y=[4,5,6,7,8]
plt.scatter(x,y)
```

```python
#pie chart
label=['A','B','C','D']
size=[20,40,60,50]
color=['yellow','green','black','white']
explode=(0.1,0,0,0)
plt.pie(size,labels=label,autopct="%1.1f%%",explode=explode)

```
