```python
df.fillna(0)#replace null val with 0
```

```python
##filling the missing val with mean
df['sales_fi']=df['Loud Cover'].fillna(df['Loud Cover'].mean())
df
```

```python
df.dtypes
```

```python
#rename vcol
df=df.rename(columns={'sales_fi;;':'sales_filled'})
df.head()
```

```python
#change datatype
df['value_new']=df['sales_filled'].astype(float)
```

```python
df['value_new']=df['sales_filled'].apply(lambda x:x*2)
df.head()
```

```python
##data aggregatigng and grouping
grouped_mean=df.groupby('Wind Speed (km/h)')['Wind Bearing (degrees)'].mean()
print(grouped_mean)
```

```python
summ=df.groupby(['Wind Speed (km/h)','Visibility (km)'])['Wind Bearing (degrees)'].sum()
print(summ)
```

```python
#aggregate multiple fxn
grouped_agg=df.groupby('Wind Speed (km/h)')['Wind Bearing (degrees)'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
#merge ing and joining dataframes
#create sample dataframe
df1=pd.DataFrame({'key':['A','B','C'],'val1':[1,2,3]})
df2=pd.DataFrame({'key':['A','B','C'],'val1':[4,5,6]})
df1
```

```python
#merge ataframe on key col
pd.merge(df1,df2,on='key',how='outer')
```

```python
pd.merge(df1,df2,on='key',how='inner')
```
