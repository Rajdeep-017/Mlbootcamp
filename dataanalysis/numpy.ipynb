```python
#attributes of numpy
arr=np.array([[1,2,3],[4,5,6]])
print("array",arr)
print("hapes",arr.shape)
print("no of dimension",arr.ndim)
print("data types",arr.dtype)
print("item size",arr.itemsize)
```

```python
#numpy vectorized operation
arr1=np.array([1,2,3,4])
arr2=np.array([6,7,8,9])
#eelement wise add
print(arr1+arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr1/arr2)
```

```python
#universal fxns
arr=np.array([2,3,4,5])
#square root
print(np.sqrt(arr))
print(np.exp(arr))
print(np.sin(arr))
print(np.log(arr))
```
