SEABORN
 Seaborn is a Python visualization library based on matplotlib.

So it is built on the top of matplotlib that provides a higher level interface for drawing attractive

and informative, uh, statistical graphics.

Seaborn helps in creating complex visualization with just a few lines of code.

```python
import seaborn as sns
```

```python
#basics
tips=sns.load_dataset('tips')
tips
```

```python
#scatter plot
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill',y='tip',data=tips)
```

```python
sns.lineplot(x='size',y='total_bill',color='blue',data=tips)
```

```python
#categorical var
sns.barplot(x='day',y='total_bill',data=tips)
```

```python
#box plot
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
#violin plot
sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
#histogram
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
#kde plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
#pairplot
sns.pairplot(tips)
```

```python
#heatmap
corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```
