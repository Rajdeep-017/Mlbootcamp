DECORATORS
 decorators are powerful and flexible feature in Python that allows you to modify the behavior of

a function or class method.

They are commonly used to add functionality to functions or methods without modifying the actual code.

```python
#decorators
def main_welcome(func):
    msg="welcome"
    def sub_welcome():
        print("welcome to adv python course")
        func()
    return sub_welcome()
```

```python
def cour_intro():
    print("this is the adv python ourse")

cour_intro()
main_welcome(cour_intro)
```

```python
#without calling the main fxn can the msgs be displayed 
#use the @ for decorator
@main_welcome
def cour_intro():
    print("this is the adv python ourse")
```

```python
#decorator
def my_dec(func):
    def wrapper():
        print("something is happen befors the func is called")
        func()
        print("something is happening aftre the func is called")
    return wrapper

@my_dec
def say():
    print("hello")
say()
```

```python
#decorator with argument
def repeat(n):
    def decor(func):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                func(*args,**kwargs)
        return wrapper
    return decor
```

```python
@repeat(2)
def say():
    print("hello")

say()
```
