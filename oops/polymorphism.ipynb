POLYMORPHISM
polymorphism is a core concept in object oriented programming that allows object of different class
to be treated as objects of a common superclass.
 it provides you a way to perform a single action in different forms.

 polymorphism is typically achieved through method overriding and interfaces.


#method overridding
 method
Overriding allows the child class to provide a specific implementation of a method that is already defined
in the parent class.

```python
#base
class ani:
    def speak(self):
        return "sound of animal"
    
#derived class
class dog(ani):
    def speak(self):
        return "woof"
    
#d2class
class cat(ani):
    def speak(self):
        return "meow"

##fxn hat demonstartes polymorphism
def ani_speak(ami):
    print(ami.speak())

dog1=dog()
cat1=cat()
print(dog1.speak())
print(cat1.speak())
ani_speak(dog1)
```

```python
##polymorphism with fxns and methods
class shape:
    def area(self):
        return " the area of the fig"

#der class
class rect(shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height

    def area(self):
        return self.width*self.height

class circle(shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius*self.radius
# fxn demonstartes ploymorphism
def print_area(shape):
    print(f"the area{shape.area()}")

rectangl=rect(4,6)
cir=circle(5)
print_area(rectangl)
print_area(cir)


```

#polymorphism with abstract base class
Abstract base class are used to define common methods for a group of related objects.

They can enforce that derived class, implement a particular method promoting consistency across the

different implementation.
