```python
#define a class eith instance method
class dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def bark(self):
         print(f"{self.name} says woof")

dog1=dog("buf",4)
dog1.bark()
dog2=dog("bb",5)
dog2.bark()    #object: diff dog type
```

```python
#modelling a Bank account
##define a class for bank acc
class bank:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited, new bbal{self.balance}")
    def withdraw(self,amount):
        if amount>self.balance:
            print("bal insuff")
        else:
            self.balance-=amount
            print(f"{amount} is withdraw new bal{self.balance}")
    
    def get_bal(self):
        return self.balance

##create acc
acc=bank("ran",10000)
print(acc.balance)
acc.deposit(1000)
```
