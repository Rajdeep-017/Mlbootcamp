INHERITANCE
inheritance is a fundamental concept in object oriented programming that allows a class to
inherit attributes and methods from another class, 

```python
class car:
    def __init__(self,windows,doors, engine):
        self.windows=windows
        self.doors=doors
        self.engine=engine
    def drive(self):
        print(f"the person {self.engine}")
    
car1=car(4,5,"petrol")
car1.drive()

```

```python
class tesla(car):
    def __init__(self, windows, doors, engine,selfdrive):
        super().__init__(windows, doors, engine)#call parent class
        self.selfdrive=selfdrive
    def slefdriving(self):
        print(f"tesla support self droveing : {self.selfdrive}")
    
tesla1=tesla(4,5,"electric",True)
tesla1.slefdriving()
tesla1.drive()
```

```python
#kultiple Inheritance
#when a class inherits from more than one base class
#base clas 1
class ani:
    def __init__(self,name):
        self.name=name
    def speak(self):
        print("sunclass miust be there")
#base class 2
class pet:
    def __init__(self,owner):
        self.owner=owner
    
#derived class
class dog(ani,pet):
    def __init__(self, name,owner):
        ani.__init__(self,name)
        pet.__init__(self,owner)
    def speak(self):
        print(f" {self.name}")

dog1=dog("jj","kri")
dog1.speak()


```
