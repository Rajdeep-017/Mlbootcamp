ENCAPSULTATION

It restricts direct accessto some of the object components, which is a mean of preventing accidental inference and misuse

```python
#ENCAPSULATION WITH  getter and setter method
##public, protected , private variables
class person:
    def __init__(self,name,age,gender):
        self.name=name    # public variable
        self.age=age       #public variable
        self.__gender=gender    #private variable
def get_name(per):
    return per.name

per=person("k",20,"M")
get_name(per)
# print(per.name)
```

```python
class person:
    def __init__(self,name,age,gender):
        self.name=name    # public variable
        self.age=age       #public variable
        self.__gender=gender    #private variable
def get_gender(per):
    return per._person__gender

per=person("k",34,"M")
get_gender(per)
```

```python
class person:
    def __init__(self,name,age,gender):
        self.name=name    # public variable
        self.age=age       #public variable
        self._gender=gender    #protected variable  #cannot acces outside the class but can acces from derived class
class employ(person):
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)

emp=employ("k",333,"M")
print(emp._gender)
```

```python
#ENCAPSULATION WITH  getter and setter
class person:
    def __init__(self,name,age):
        self.__name=name##private access modifier or varible
        self.__age=age#private var

    #getter method for name
    def get_name(self):
        return self.__name
    #setter method for name
    def set_name(self,name):
        self.__name=name
    #getter age
    def get_age(self):
        return self.__age
    #setter age
    def set_age(self,age):
        if age>0:
            self.__age=age
        else:
            print("age cannot be negative")

per1=person('kkrr',32)

##acces and modify private var getter and setter
print(per1.get_name())
print(per1.get_age())
per1.set_age(444)
print(per1.get_age())
```
