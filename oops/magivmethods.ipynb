```python
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self):
        return f"{self.name},{self.age} years old"
    
    def __repr__(self):
        return f"person(name={self.name},age={self.age})"
    
per=person("krish",24)
print(per)
print(repr(per))
```
