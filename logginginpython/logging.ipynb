PYTHON LOGGING
logging is crucial aspects of any application

providing ways to track event errors and operational information.

Python built in logging module offers a flexible framework for emitting log messages for Python program.

In this lesson, we'll cover the basics of logging, including how to configure logging, logging levels,

and the best practices for using logging in Python application.

```python
import logging
#configure the basic logging setting
logging.basicConfig(level=logging.DEBUG)
logging.debug('this is a debug meg')
logging.info('this is an info meg')
logging.warning("warning msg")
logging.error("this is error")
logging.critical('critical msg')
```

Log level
DEBUG:debug detailed information, typically of interest only when diagnosis problem.
INFO:confirmation that things are working as expected.
WARNING:Indication that something unexpected

happened or indicative of some problem in the near future.
ERROR:due to your most serious problem, the software has not been able to perform some

function and critical.
CRITICAL:A very serious error indicating the program itself may not be, uh, may be unable to continue running.

```python
#congifuring the logging
import logging
logging.basicConfig(
    level=logging.DEBUG,
    filename='app.log',
    filemode='w',
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datafmt='%Y-%m-%d %H:%M:%S'
)
#configure the basic logging setting
logging.basicConfig(level=logging.DEBUG)
logging.debug('this is a debug meg')
logging.info('this is an info meg')
logging.warning("warning msg")
logging.error("this is error")
logging.critical('critical msg')
```
