```python
import logging
#create logger for module1
logger1=logging.getLogger('module1')
logger1.setLevel(logging.DEBUG)

#module2
logger2=logging.getLogger("module2")
logger2.setLevel(logging.WARNING)
#config logging setting
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

```python
#log messages 
logger1.debug('this is a debug meg')
logger2.info('this is an info meg')
logger1.warning("warning msg")
logger2.error("this is error")
```
