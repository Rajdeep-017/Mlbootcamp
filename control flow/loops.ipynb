```python
# for
for i in range(0,6):
    print(i)
```

```python
for i in range(1,6):
    print(i)
for i in range(1,10,2):
    print(i)
for i in range(10,0,-1):
    print(i)
```

```python
#string
str="ran"
for i in str:
    print(i)
```

```python
# while loop
# continue wntill the condition is true
count =0
while(count<5):
    print(count)
    count+=1
```

```python
cnt=0
while cnt%2==0:
    print(cnt)
    cnt+=1
```

```python
#loop controls
#break
# exits the loop when the condition is true
for i in range(1,10):
    if i==5:
        break
    print(i)
```

```python
#continue
# the continue statemet skips the current iteration and continues with the next one
for i in range(1,10):
    if i==5:
        continue
    print(i)
```

```python
#PASS
# the pass statement is used when a statement is required syntactically but you do not want any command or code to execute
#pass statement is a null statement in python
for i in range(1,10):
    if i==5:
        pass
    print(i)
```

```python
#nested loops
# a loop inside another loop is called nested loop
for i in range(1,4):
    for j in range(1,4):
        print(i,j)
```

```python
# ex calu the sum of n natural no with while loop
n=10
sum=0
i=1
while i<=n:
    sum+=i
    i+=1
print("The sum of first", n, "natural numbers is:", sum)
   
```

```python
n=10
sum=0
for i in range(11):
    sum+=i
print("The sum of first", n, "natural numbers is:", sum) 
```
