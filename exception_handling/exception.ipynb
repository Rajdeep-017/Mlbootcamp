Exception handling in Python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program.
keywords like try except else and finally blocks.

what are exception
Exceptions are events that disrupt the normal flow of a program.

They occur where an error is encountered during program execution.

Common exception include zero division error.

So here we have something like divide by zero file not found or value error type error.

These are some of the example okay.



```python
#exception handlin try,except,block
try:
    a=b
except:
    print("the variable is not assigned")
```

```python
try:
    a=b
except NameError as ex:
    print(ex)
```

```python
try:
    res=1/0
except ZeroDivisionError as ex:
    print(ex)
    print("enter deno more than zero")
```

```python
try:
    res=1/0
    a=b
except ZeroDivisionError as ex:
    print(ex)
    print("enter deno more than zero")
except Exception as exp: # base exception can handle any exception
    print(exp)
    print("main exception got caught here")
```

```python
## try, except ,else block
try: 
    num(int(input("enter")))
    res=10/num
except ValueError:
    print("not a valid no")
except ZeroDivisionError:
    print("enter deno greater than 0")
except Exception as ex:
    print(ex)
else:
    print("the res is {res}")
```

```python
#try ,except,else ,finally
try: 
    num=int(input("enter"))
    res=10/num
except ValueError:
    print("not a valid no")
except ZeroDivisionError:
    print("enter deno greater than 0")
except Exception as ex:
    print(ex)
else:
    print(f"the res is {res}")
finally:
    print("execution completed")
```

```python
#practical ex
#file handlinfg and exception hanl
try:
    file=open('example1.txt','r')
    content=file.read()
    print(content)
except FileNotFoundError:
    print("file not found")
except Exception as ex:
    print(ex)
finally:
    if 'file' in locals() and not file.closed():
        file.close()
        print("closed!!!")
```
