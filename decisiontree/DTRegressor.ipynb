```python
from sklearn.datasets import load_diabetes
dataset=load_diabetes()
```

```python
dataset
```

```python
import pandas as pd
df= pd.DataFrame(dataset.data, columns=dataset.feature_names)
```

```python
df
```

```python
x=df
y=dataset['target']
```

```python
from sklearn.model_selection import train_test_split
x_t,x_test,y_t,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
x_t.head()
```

```python
x_t.corr()
```

```python
import seaborn as sns
sns.heatmap(x_t.corr())
```

```python
from sklearn.tree import DecisionTreeRegressor
reg=DecisionTreeRegressor()
reg.fit(x_t,y_t)
```

```python
param={
    'criterion':['squared_error','fried,ax_mse','absolute_error'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['auto','sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
reg=DecisionTreeRegressor()
grid=GridSearchCV(reg,param_grid=param,cv=5,scoring='neg_mean_squared_error')
```

```python
grid.fit(x_t,y_t)
```

```python
grid.best_score_
```

```python
y_pred=grid.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,r2_score
print(r2_score(y_pred,y_test))
print(mean_squared_error(y_pred,y_test))
```
