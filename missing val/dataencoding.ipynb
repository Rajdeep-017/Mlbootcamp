Nominal/OHE encoding
One hot encoding, also known as nominal encoding, is a technique used to represent categorical data as a numerical data which is more suitable for machine learning algorithms.
In this technique, each category is represented as a binary vector, where each bit corresponds to
a unique category.

For example, if we have a categorical variable color with three possible values like red, green,

blue, then we can represent this using one hot encoding like 100010001.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
# create dataframe
df=pd.DataFrame({
    'color':['red','blue','green','red','blue']
})
```

```python
df.head()
```

```python
#create the instance of encoder
encoder=OneHotEncoder()
```

```python
encoded=encoder.fit_transform(df[['color']]).toarray()
```

```python
import pandas as pd
encoded_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoded_df
```

```python
#for new data
encoder.transform([['blue']]).toarray()
```

LabelEncoding
Label encoding and ordinal encoding are two techniques used to encode categorical

data as numerical data.

Label encoding involves assigning a unique numerical label to each category in the variable.
for ex:
Red=1,
green=2
blue=3

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder=LabelEncoder()
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['red']])
```

ordinal encoding  assign ranks

For example, if we have a categorical variable education level with four possible values like high

school, college graduate, and post graduate, we can represent it using ordinal encoding.
high school=1

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
    'size':['small','medium','large','medium','large','small']
    
})
```

```python
df
```

```python
od_encoder.fit_transform(df[['size']])
```

```python
od_encoder.transform([['small']])
```

TargetGuided Ordinal Encoding
It is a technique used to encode categorical variable based on the relationship with the target variable. This encoding technique is useful when we have categorical variable with a large number of unique categories.

we replace each category in the categorical variable

with the numerical value based on the mean or median of the target variable for that category.


```python
import pandas as pd
df=pd.DataFrame({
    'city':['nyc','london','paris','tokyo','nyc','paris'],
    'price':[200,400,800,250,320,350]
})
```

```python
df
```

```python
mean_price=df.groupby('city')['price'].mean()
```

```python
df['city_encoded']=df['city'].map(mean_price)
df
```

```python
df[['price','city_encoded']]
```
