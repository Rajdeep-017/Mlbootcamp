```python
import pandas as pd
import numpy as np
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
df['Gender']=df['Gender'].replace('Fe Male','Female')
df['maritalStatus']=df['MaritalStatus'].replace('Single','unmarried')
```

```python
df.head()
```

```python
df['Gender'].value_counts()
```

```python
#missing values

feature_with_na=[features for features in df.columns if df[features].isnull().sum()>=1]
for feature in feature_with_na:
    print(feature,np.round(df[feature].isnull().mean()*100,5),'% missing values')
```

```python
df[feature_with_na].select_dtypes(exclude='object').describe()
```

```python
df.Age.fillna(df.Age.median(),inplace=True)
df.TypeofContact.fillna(df.TypeofContact.mode()[0],inplace=True)
df.DurationOfPitch.fillna(df.DurationOfPitch.median(),inplace=True)
df.NumberOfFollowups.fillna(df.NumberOfFollowups.mode()[0],inplace=True)
df.PreferredPropertyStar.fillna(df.PreferredPropertyStar.mode()[0],inplace=True)
df.NumberOfTrips.fillna(0,inplace=True)
df.NumberOfChildrenVisiting.fillna(df.NumberOfChildrenVisiting.mode()[0],inplace=True)
df.MonthlyIncome.fillna(df.MonthlyIncome.median(),inplace=True)
```

```python
df.head()
```

```python
df.drop('CustomerID',inplace=True,axis=1)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df['TotalVisiting']=df['NumberOfPersonVisiting']+df['NumberOfFollowups']
df.drop(columns=['NumberOfPersonVisiting','NumberOfFollowups'])
```

```python
num_features=[feature for feature in df.columns if df[feature].dtype!='O']
print("no of numerical feature", len(num_features))
```

```python
cat_features=[feature for feature in df.columns if df[feature].dtype=='O']
print("no of numerical feature", len(cat_features))
```

```python
dis_features=[feature for feature in df.columns if len(df[feature].unique())<=25]
print("no of numerical feature", len(dis_features))
```

```python
#continuoues featuere
cont_features=[feature for feature in num_features if feature not in dis_features]
print("no of numerical feature", len(cont_features))
```

```python
from sklearn.model_selection import train_test_split
x=df.drop(['ProdTaken'],axis=1)
y=df['ProdTaken']
```

```python
x_t,x_test,y_t,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
x_t.shape
```

```python
cat_features=x.select_dtypes(include='object').columns
num_features=x.select_dtypes(include='object').columns
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
numeric_transformer=StandardScaler()
oh_transformer=OneHotEncoder(drop='first')
preprocess=ColumnTransformer(
    [
    ('OneHotEncoder',oh_transformer,cat_features),
    ('StandardScaler',numeric_transformer,num_features)
    ]
)
```

```python
preprocess
```

```python
x_t.dtypes
```

```python
#applying transformation in training(fit_transform)
x_t=preprocess.fit_transform(x_t)
```

```python
x_t
```

```python
x_test=preprocess.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,r2_score,recall_score,precision_score,f1_score,roc_auc_score,classification_report,confusion_matrix
```

```python
models={
    'Randon Forest':RandomForestClassifier(),
    'Decision Tree':DecisionTreeClassifier()

}
for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(x_t,y_t)
    y_train_pred=model.predict(x_t)
    y_test_pred=model.predict(x_test)
    model_train_acc=accuracy_score(y_t,y_train_pred)
    model_train_f1=f1_score(y_t,y_train_pred,average='weighted')
    model_train_precision=precision_score(y_t,y_train_pred)
    model_train_recall=recall_score(y_t,y_train_pred)
    model_train_rocauc=roc_auc_score(y_t,y_train_pred)
    #test data
    model_test_acc=accuracy_score(y_t,y_test_pred)
    model_test_f1=f1_score(y_t,y_test_pred,average='weighted')
    model_test_precision=precision_score(y_t,y_test_pred)
    model_test_recall=recall_score(y_t,y_test_pred)
    model_test_rocauc=roc_auc_score(y_t,y_test_pred)
    print(list(models.key())[i])
    print("accuracy".format(model_train_acc))
    print("f1_score".format(model_train_f1))
    print("precision_score".format(model_train_precision))
    print("recall_score".format(model_train_recall))
    print("rocauc_score".format(model_train_rocauc))
    #test
    print("test acc".format(model_test_acc))
    print("f1_score".format(model_test_f1))
    print("precision_score".format(model_test_precision))
    print("recall_score".format(model_test_recall))
    print("rocauc_score".format(model_test_rocauc))
```
