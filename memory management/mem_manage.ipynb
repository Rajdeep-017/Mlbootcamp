reference counting
it is a primary method Python uses to manage memory.
Each object in Python maintains a count of references pointing to it.
When the references count drops to zero, the memory occupied by the object is deallocated.

```python
#reference counting
import sys
a=[]
# 2(one reference from 'a) and one from getrefcount()
print(sys.getrefcount(a))

```

```python
b=a
print(sys.getrefcount(b))
```

```python
del b
print(sys.getrefcount(a))
```

#garbage collection
 python includes a cyclic garbage collector to handle reference cycles.

Reference cycles occurs when object reference each other, preventing the reference counts from reaching zero


```python
import gc
#enable garbage collection
gc.enable()
```

```python
gc.disable()
```

```python
gc.collect()
```

```python
#get garbage collection stats
print(gc.get_stats())
```

```python
#get unreachable garbage
print(gc.garbage)
```

memory management best practice
local variables:because local variable have a shorter lifespan and freed sooner then global variables.
Avoid circular reference.:circular reference can lead to memory leaks.Circular reference basically means a is equal to b, b is equal to a.
USE GENERATORS: generator produce items one at a time and only keep one item in memory
Explicitly delete objects:delete statement to delete variable and objects explicitly whenever it is not used.
profile memory usage:memory profiling tools like trace, malloc and Memory Profiler to identify memory

```python
#handling circular reference
import gc
class Myobj:
    def __init__(self,name):
        self.name=name
        print("objects created",self.name)
    def __del__(self):
        print(f"object {self.name} deleted")

obj1=Myobj('obj1')
obj2=Myobj('obj2')
obj1.ref=obj2
obj2.ref=obj1

del obj1
del obj2

#manually trigger the garbage collection
gc.collect()
```

```python
#generator for memory efficiency
#generator allows you to produce one item at one time using memory efficiently by only keeping one item in the memory and removing that particular time.

def gen_num(n):
    for i in range(n):
        yield i
#using the generator
for num in gen_num(100000):
    print(num)
    if num>10:
        break
```

```python
#profiling memory usage with tracemalloc
import tracemalloc
def create():
    return [i for i in range(10000)]
def main():
    tracemalloc.start()
    create()
    snap=tracemalloc.take_snapshot()
    top=snap.statistics('lineno')
    print('[top 9]')
    for stat in top[::]:
        print(stat)
```

```python
main()
```
