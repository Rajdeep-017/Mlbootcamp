```python
#multi corlinearity the independent features are highly related 
def correlation(dataset,threshold):
    col_corr=set()
    corr_matrix=dataset.corr()
    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if abs(corr_matrix.iloc[i,j])> threshold:
                colname=corr_matrix.columns[i]
                col_corr.add(colname)
    return col_corr

    
```

```python
#threshold ==domain exppert
corr_features=correlation(x_t,0.85)
```

```python
corr_features
```

```python
#drop features when corrleation is more thaan 0.85
x_t.drop(corr_features,axis=1,inplace=True)
x_test.drop(corr_features,axis=1,inplace=True)
x_t.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_t_s=scaler.fit_transform(x_t)
x_test_s=scaler.transform(x_test)
```

```python
plt.subplots(figsize=(15,5))
plt.subplot(1,2,1)
sns.boxplot(data=x_t)
plt.subplot(1,2,2)
sns.boxplot(data=x_t_s)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,r2_score
linear=LinearRegression()
linear.fit(x_t_s,y_t)
y_pred=linear.predict(x_test_s)
mae=mean_absolute_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
plt.scatter(y_test,y_pred)
```

```python
from sklearn.linear_model import Lasso
lasso=Lasso()
lasso.fit(x_t_s,y_t)
y_pred=lasso.predict(x_test_s)
mae=mean_absolute_error(y_test,y_pred)
plt.scatter(y_test,y_pred)
```

```python
from sklearn.linear_model import LassoCV
lassocv=LassoCV(cv=5)
lassocv.fit(x_t_s,y_t)
lassocv.alpha_
```

```python
lassocv.alphas_
```

```python
y_pred=lassocv.predict(x_test_s)
plt.scatter(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
print(mae)
```

```python
from sklearn.linear_model import Ridge
ridge=Ridge()
ridge.fit(x_t_s,y_t)
y_pred=ridge.predict(x_test_s)
mae=mean_absolute_error(y_pred,y_test)
score=r2_score(y_test,y_pred)
print(mae)
print(score)
plt.scatter(y_test,y_pred)
```

```python
from sklearn.linear_model import RidgeCV
ridgecv=RidgeCV()
ridgecv.fit(x_t_s,y_t)
y_pred=ridgecv.predict(x_test_s)
mae=mean_absolute_error(y_pred,y_test)
score=r2_score(y_test,y_pred)
print(mae)
print(score)
plt.scatter(y_test,y_pred)
```
