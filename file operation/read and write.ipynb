FILE OPERATIONS read and write operations
file handling is  a crucial part od any programminf lang python provides built-in fxns and methods to read from and write to files both text and binary


```python
#read whole file
with open("example.txt",'r') as file:
    content=file.read()
    print(content)
     
```

```python
#read the file line by line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) # removes the  newline char
```

```python
##writing in file(overwriting)
with open('example.txt','w') as file:
    file.write("hello Uk \n")
    file.write("pepe offf")
 
 # now the entire content gets over written
 
```

```python
#write a file with out overwriting
with open("example.txt",'a') as file: #a=appended
    file.write("\nAppednd operation taking place\n")
```

```python
#write multiple lines
lines =["apple \n","mago \n","third \n"]
with open("example.txt",'a') as file:
    file.writelines(lines)
```

```python
#binary files
#writing 
data=b'\x00\x01\x02'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
#reading the binary file
with open('example.bin','rb') as file:
    content=file.read()
    print(content)
```

```python
#read the content from source text file and write to the destination text file

with open("example.txt",'r') as source_file:
    content=source_file.read()

with open ('destination.txt','w') as destination_file:
    destination_file.write(content)
```

```python
#read the textfile anf count the number of lines words and char
#counting lines words and char in text file
def count_text(file_path):
    with open(file_path,'r') as file:
        lines=file.readlines()
        line_count=len(lines)
        word_count=sum(len(line.split()) for line in lines)
        char_count=sum(len(line) for line in lines)
    return line_count,word_count,char_count
file_path='example.txt'
lines,words,char=count_text(file_path)
print(f'lines: {lines}, words: {words}, chars:{char}')
```

The W plus mode in Python is used to open a file for both reading and writing.If the file does not exist, it will be created.If the file exists, the its content is truncated.That is, the file is overwritten.

```python
#writing and reading the file
with open('example.txt','w+') as file:
    file.write('hello world\n')
    file.write('this is the new line')
    #move the file cursor to the begin
    file.seek(0)
    #read te content of the file
    content=file.read()
    print(content)
```
