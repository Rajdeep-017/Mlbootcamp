```python
#variable len argument
#positional and keyword args
#may have multi parameters
def printn(*args):
    for n in args:
        print(n)

printn(1,2,3,4,5,6)
```

```python
##keywords arguments
def printd(**kwargs):
    for key,val in kwargs.items():
        print(f"{key}:{val}")

printd(name='raj', age=324)
```

```python
def printd(*args,**kwargs):
    for val in args:
        print(val)
    for key,val in kwargs.items():
        print(f"{key}:{val}")

printd(1,2,3,name='raj', age=324)
```
