#LAMBDA
Lambda functions are small anonymous function defined using the lambda keyword.
they can have any number of arguments, uh, but only one expression.
And they are commonly used for short operation or as arguments to higher order function.

```python
def add(a,b):
    return a+b
add(2,6)
```

```python
add=lambda a,b:a+b
add(2,3)
```

```python
def even(num):
    if num%2==0:
        return True

even(24)
```

```python
even1=lambda num:num%2==0
even1(24)
```

```python
def add(x,y,z):
    return x+y+z
add(1,3,5)
```

```python
add1=lambda x,y,z:x+y+z
add1(3,4,9)
```

```python
#map-applies a fxn to all items in the list
num=[1,2,3,4,5]
def sq(num):
    return num**2
```
