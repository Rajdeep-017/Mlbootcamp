FILTER
 filter function constructs an iterator from elements of an iterable for which a function returns
true.

It is used to filter out items from a list based on a condition.

```python
def evem(num):
    if num%2==0:
        return True
evem(20)
```

```python
lst=[1,2,3,4,5,6,7]
list(filter(evem,lst))
```

```python
##filters with lambda
num=[1,2,3,4,6]
greater=list(filter(lambda x:x>5, num))
```

```python
#filter with lambda fxn and multiple condition
num=[1,2,3,4,5,6,7]
evenand_great=list(filter(lambda x:x>5 and x%2==0, num))
```

```python
#filter to check the age 


people=[
    {'name':'krish','age':32},
    {'name':"jac",'age':22}
]


def age_g(pers):
    return pers['age']>25
list(filter(age_g,people))
```

filter function is a powerful tool for creating iterators that filter items

out of an iterable based on a function.

It is commonly used for data cleaning filter objects and removing unwanted elements.

By mastering filter, you can write more concise and efficient code for processing and manipulating

collections in Python.
