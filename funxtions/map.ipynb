MAP
It applies a given function to all the items in an input list or any other iterable, and returns a

map object an iterator.
This is particularly useful for transforming data in list comprehensively.

```python
def sq(x):
 return x*x
sq(10)
```

```python
num=[1,2,3,4,5,6]
list(map(sq,num))
```

```python
#lambda
list(map(lambda x:x*2,num))
```

```python
num1=[1,2,3]
num2=[4,5,6]
added=list(map(lambda x,y : x+y,num1,num2))
```

```python
##map() to convert str to int
# use map
str_num=['1','2','3','4']
int_num=list(map(int,str_num))
print(int_num)
```

```python
word=['apple','mango','cherry']
upper_word=list(map(str.upper,word))
```

```python
def get(pers):
    return pers['name']

people=[
    {'name':'krish','age':32},
    {'name':"jac",'age':22}
]
list(map(get,people))
```
