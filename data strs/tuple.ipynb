TUPLE
tuples are ordered collection of items that are immutable.

```python
##creating
tup=()
print(tup)
print(type(tup))
```

```python
num=tuple([1,2,3,4])
print(num)
print(type(num))
```

```python
list=((1,2,3,4))
print(list)
print(type(list))
```

```python
mixed=(1,2,3,"a","b","c")
print(mixed)
```

```python
## accessing tuple element
mixed[0]
print(mixed[-1])
```

```python
mixed[::-1]
```

```python
##tuple operations
tup1=(1,2,3)
tup2=(4,5,"hello")
tup3=tup1+tup2
print(tup3)
```

```python
mixed*3
```

```python
##immutable nature of tuple 
# means their elements cannot be changed after creation
lst=[1,2,3,4]

print(lst)
lst[0]=100 
print(lst)

#list is mutable
tup=(1,2,3,4)
tup[0]=100
print(tup)
```

```python
print(tup.count(1))
print(tup.index(3))
```

```python
#packing and unpacking
pac="hell",1,3.122
print(pac)
#unpacking
a,b,c=pac
print(a,b,c)

```

```python
#unpacking with *
num=(1,2,3,4,5,6)
a,*b,c=num
print(a)
print(b)
print(c)
```

```python
##nested tuple
lst=[[23,4],[2,3,5],["hello",1,2,4]]
print(lst)
lst[0][0:1]
```

```python
lst=[[23,4],[2,3,5],("hello",1,2,4)]
lst[2][0:2]
```

```python
nest_tup=((1,2,3),(4,5,6),("hello",1,2,3))
print(nest_tup)
nest_tup[0]
nest_tup[2][2]
```

```python
for nes in nest_tup:
    for item in nes:
        print(item)
    print()
```

tuples are tuples are versatile and very useful in various, uh, in many real world scenarios where

arguments, and return values as dictionary keys.
