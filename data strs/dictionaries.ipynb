DICTIONARIES

 dictionaries are nothing, but it is an unordered collection of items.

The store data in key value pairs Key must be unique and immutable.

```python
##creating 
dict={}
print(type(dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"krish","age ":23,"grad":12}
print(student)
```

```python
#error (key should be unique)
student={"name":"krish","age":23,"name":12}
print(student)
```

```python
#acccesing dict element
student={"name":"krish","age ":23,"grad":12}
print(student)
student["grad"]
student["age "]

#accesing using get() method
print(student.get('grad'))
print(student.get('age '))
print(student.get('last','not available'))
```

```python
## modifying dict ele
##dict are mutable ,so you can add or delete ele
print(student)
```

```python
student["age "]=33 ## update value
print(student)
student["address"]="india"  # added new val
print(student)
```

```python
del student["grad"]
print(student)
```

```python
#dict methods
keys=student.keys() #get all keys
print(keys)
value=student.values() # get all val
print(value)
items=student.items() #get all keys value
print(items)
```

```python
#shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="raj"
print(student)
print(student_copy)
student_copy1=student.copy() # shallow copy
print(student)
print(student_copy1)
```

```python
student["name"]="ram"
print(student)
print(student_copy1)
```

```python
##iterating over dict
##loops
#iterating over keys
for key in student.keys():
    print(key)
for keys in student.values():
    print(keys)
#itera key values pairs
for key,keys in student.items():
    print(f"{key}:{keys}")

```

```python
#nested dictionaries
students={"std1":{"name":"ram","age":10},
          "std2":{"name":"raj","age":20}}
print(students)
```

```python
print(students["std2"]["name"])
```

```python
#iteration in nested dict
for stdid, stdinfo in students.items():
    print(f"{stdid}:{stdinfo}")
    for key ,value in stdinfo.items():
        print(f"{key}: {value}")
```

```python
#dict comprehension
sq={ x:x**2 for x in range(3)}
print(sq)
```

```python
##conditional dic coprehen
evens={x: x**2 for x in range(6) if x%2==0}
print(evens)
```

```python
#practicla ex
#use dict to count the freq of ele in list
num=[1,2,3,4,5,6,4,5,6]
freq={}
for n in num:
    if n in freq:
        freq[n]+=1
    else: freq[n]=1
print(freq)
```

```python
#3merge two dict
dict1={"a":1,"b":3}
dict2={"b":9,"c":4}
merged={**dict1,**dict2}
print(merged)
```
