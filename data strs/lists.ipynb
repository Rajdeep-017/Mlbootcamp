## list comprehension with condition
basics syn [expression for item in iterable]
with conditional logic  [exp for item in iterable if condition]
nested list comprehension [exp for item1 in iterable1 for item2 in iterable2]

```python
## baisc list operations
sq=[n**2 for n in range(10)]
```

```python
lst=[]
for i in range(5):
    if i%2==0:
        lst.append(i)
print(lst)
```

```python
even=[i for i in range(5) if i%2==0]
```

```python
#nested list
l1=[1,2,3,4]
l2=['a','b','c','d']
pair=[(i,j) for i in l1 for j in l2]
print(pair)
```

```python
#list compre with fxn call
word=["help","me","out"]
leng=[len(w) for w in word]
print(leng)
```
