sets are built in data types in Python used to store collection of unique

items.

They are unordered meaning that the elements do not follow a specific order and they do not allow duplicate

elements.
 sets are useful for

membership test.

Eliminating duplicate entries.

Performing mathematical set operations like union intersection differences and semantic differences.

```python
#creating 
set1={1,2,3}
print(type(set1))

```

```python
set2=set([1,2,3,4])
print(set2)
```

```python
#no duplicates
set2=set([1,2,2,4,3,4])
print(set2)
```

```python
#basic set operations
#add and 
# remove ele(if ele not present gives error)
set2.add(12)

print(set2)
set2.remove(3)
print(set2)

```

```python
set2.discard(10)
print(set2)
#pop method
r=set2.pop()
print(r)
print(set2)
```

```python
##clear all ele
set2.clear()
```

```python
##set membership test
set3={1,2,3,4,5}
print(3 in set3)
print(10 in set3)
```

```python
##mathematical operation
set4={1,2,3,4,5,6,7}
set5={7,8,9,10}
unionset=set4.union(set5)
print(unionset)
#intersection
inter=set4.intersection(set5)
print(inter)
set4.intersection_update(set5)
print(set4)
```

```python
set4={1,2,3,4,5,6,7}
set5={7,8,9,10}
#difference
print(set4.difference(set5))
set4.difference_update(set5)
print(set4)
```

```python
##symmetric diff
set4={1,2,3,4,5,6,7}
set5={7,8,9,10}
print(set4.symmetric_difference(set5))
```

```python
#sets methods
set1={1,2,3,4,5}
set2={3,4,5}
#is_subset
print(set1.issubset(set2))
set1.issuperset(set2)
```

```python
lt=[1,2,3,4]
set(lt)
```

```python
##count unique word in text
text="Arithmetic tutorial and the classes"
word=text.split()
uniqu=set(word)
print(uniqu)
print(len(uniqu))
```
