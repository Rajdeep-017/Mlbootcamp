```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#synthetic data pts
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_samples=1000,n_features=2,n_classes=2,n_clusters_per_class=1,n_redundant=0)
```

```python
# first colum
df=pd.DataFrame(x)
```

```python
df = pd.DataFrame(x, columns=["Feature1", "Feature2"])

sns.scatterplot(data=df, x="Feature1", y="Feature2", hue=y)
```

```python
sns.scatterplot(x=df[0],y=df[1])
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
x_t,x_test,y_t,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
svc=SVC(kernel='linear')
```

```python
svc.fit(x_t,y_t)
```

```python
y_pred=svc.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix

print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
rbf=SVC(kernel='rbf')
```

```python
rbf.fit(x_t,y_t)
```

```python
y_pred=rbf.predict(x_test)
```

```python
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
polynomial=SVC(kernel='poly')
polynomial.fit(x_t,y_t)
y_pred2=polynomial.predict(x_test)
print(classification_report(y_test,y_pred2))
print(confusion_matrix(y_test,y_pred2))
```

```python
sigmoid=SVC(kernel='sigmoid')
sigmoid.fit(x_t,y_t)
y_pred2=sigmoid.predict(x_test)
print(classification_report(y_test,y_pred2))
print(confusion_matrix(y_test,y_pred2))
```

hyperparameter tuning SVC

```python
from sklearn.model_selection import GridSearchCV
param_grid={'C':[0.1,1,10,100,1000],
            'gamma':[1,0.1,0.01,0.001,0.0001],
            'kernel':['rbf']
            }
```

```python
grid=GridSearchCV(SVC(),param_grid=param_grid,refit=True,cv=5,verbose=3)
```

```python
grid.fit(x_t,y_t)
```
