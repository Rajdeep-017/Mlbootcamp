```python
x=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df['total_bill']
```

```python
x_t,x_test,y_t,y_test=train_test_split(x,y,test_size=0.25,random_state=0)
```

```python
x_t.head()
```

```python
#feature encoding (labek and onehot)
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
x_t['sex']=le1.fit_transform(x_t['sex'])
x_t['smoker']=le2.fit_transform(x_t['smoker'])
x_t['time']=le3.fit_transform(x_t['time'])
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
onehot=OneHotEncoder()
ct=ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3])],remainder='passthrough')
```

```python
import sys
np.set_printoptions(threshold=sys.maxsize)
x_t=ct.fit_transform(x_t)
```

```python
x_test=ct.fit_transform(x_test)
```

```python
from sklearn.svm import SVR
svr=SVR()
svr.fit(x_t,y_t)
```

```python
y_pred=svr.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,y_pred))
```

```python
from sklearn.model_selection import GridSearchCV
param_grid={'C':[0.1,1,10,100,1000],
            'gamma':[1,0.1,0.01,0.001,0.0001],
            'kernel':['rbf']
            }
```

```python
grid=GridSearchCV(SVR(),param_grid,refit=True,verbose=3)
```

```python
grid.best_params_
```
