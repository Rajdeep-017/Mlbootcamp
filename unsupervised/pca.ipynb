```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

```

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer=load_breast_cancer()
```

```python
df=pd.DataFrame(cancer.data, columns=cancer.feature_names).head()
```

```python
x=cancer.data
y=cancer.target
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
scaler.fit(x)
```

```python
sc_data=scaler.transform(x)
```

```python
##apply pCA algo
from sklearn.decomposition import PCA
```

```python
pca=PCA(n_components=2)
```

```python
data_pca=pca.fit_transform(sc_data)
data_pca
```

```python
pca.explained_variance_
```

```python
plt.scatter(data_pca[:,0],data_pca[:,1],c=y,cmap='viridis')
plt.xlabel('first pricipal component')
plt.ylabel('second principal component')
```
