```python
from sklearn.datasets import make_classification
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_curve,roc_auc_score
```

```python
from sklearn.linear_model import LogisticRegression
x,y=make_classification(n_samples=1000,n_classes=2,random_state=1)
```

```python
from sklearn.model_selection import train_test_split
x_t,x_test,y_t,y_test=train_test_split(x,y,test_size=0.25,random_state=0)
```

```python
#create the dummy model (majority class)
dummy_model_prob=[0 for _ in range(len(y_test))]
dummy_model_prob
```

```python
model=LogisticRegression()
model.fit(x_t,y_t)
```

```python
#prediction based on probab
model_prob=model.predict_proba(x_test)
#focus baesd on positive  outcome
model_prob=model_prob[:,1]
```

```python
dummy_model_auc=roc_auc_score(y_test,dummy_model_prob)
model_auc=roc_auc_score(y_test,model_prob)
print(dummy_model_auc)
print(model_auc)
```

ROC AND AUC gives false positive rate

True postive rate
=tp/p=tp/tp+fn=1-fnr


```python
#caluc ROC curve
dummy_fpr,dummy_tpr,_=roc_curve(y_test,dummy_model_prob)
model_fpr,model_tpr,thresholds=roc_curve(y_test,model_prob)
```

```python
plt.plot(dummy_fpr,dummy_tpr,linestyle="--",label='DUMMY MODEL')
plt.plot(model_fpr,model_tpr,marker='.',label="lagistic")
plt.xlabel("false positve")
plt.ylabel("true positive")
plt.legend()
plt.show()
```

```python
fig=plt.figure(figsize=(20,50))
plt.plot(dummy_fpr,dummy_tpr,linestyle="--",label='DUMMY MODEL')
plt.plot(model_fpr,model_tpr,marker='.',label="lagistic")
ax=fig.add_subplot(111)
for xyz in zip(model_fpr,model_tpr,thresholds):
    ax.annotate('%s' % np.round(xyz[2],2),xy=(xyz[0],xyz[1]))
plt.xlabel("false positve")
plt.ylabel("true positive")
plt.legend()
plt.show()
#threshold curve where the tp is maxi and the fp is low here the threshold is 0.5 where the val beolow 0.5 is 0 and 1 above
```
